### 2.2. Kiến trúc — mục 5.1, trang PDF 13–14

Baseline chọn một tay chính, lấy x/y của 21 landmark, tạo **42 số/frame**.
Mạng gồm **ba LSTM một chiều, mỗi tầng 256 hidden units**, dropout giữa
các tầng và fully connected từ trạng thái cuối đến **14 logits**.

```text
[B, T, 42] → LSTM ba tầng, hidden 256 → [B, 256] → Linear → [B, 14]
```

### 2.3. Thí nghiệm — mục 5.2, trang PDF 14

Bài chia 50 người thành **37 train / 13 test**, tương ứng 148/52 video.
Thí nghiệm phân loại 14 lớp, loại các đoạn No-Gesture. Chuỗi dài chia
cửa sổ tối đa 64 frame, stride 32; chuỗi ngắn giữ độ dài, không padding
hoặc nội suy. Dùng Adam, learning rate 0,001, batch 64, 30 epoch,
dropout 0,3 và ba lần chạy.

### 2.4. Kết quả gốc — Bảng 6, trang PDF 15



*Nguồn ảnh — Benitez-Garcia et al., Applied Sciences 15(11), 6321, 2025,

## 4. Đọc hiểu code đồ án theo đường đi dữ liệu

```text
Clip → MediaPipe → 21 landmark xy
     → extract_features → vector 44
     → SequenceBuffer → [30,44]
     → GestureLSTM → [1,5] logits
     → softmax → nhãn và độ tin cậy
```

#### Kiểm tra bộ dữ liệu hiện có

````python
datasets = {}
sessions = {}
counts_by_split = {}

for split in ("train", "validation", "test"):
    features, targets, rows = load_split(config, split)
    datasets[split] = (features, targets, rows)
    sessions[split] = {row["session_id"] for row in rows}
    counts_by_split[split] = np.bincount(targets, minlength=len(CLASS_NAMES))
    print(
        f"{split:10} | {len(targets):3} clip | shape={features.shape} | "
        f"phiên={', '.join(sorted(sessions[split]))}"
    )

for first, second in (("train", "validation"), ("train", "test"), ("validation", "test")):
    if sessions[first] & sessions[second]:
        raise ValueError(f"Phiên thu bị trùng giữa {first} và {second}.")
````

### 4.2. Trích đặc trưng và ghép chuỗi

Chọn một clip **thuộc tập train** để minh họa. Mỗi frame có 21 điểm x/y.
`extract_features()` đưa hai trục về cùng đơn vị ảnh, trừ tọa độ cổ tay và
chuẩn hóa theo kích thước lòng bàn tay. Vector cuối gồm **42 tọa độ cục bộ**
và **2 tọa độ cổ tay toàn cục**, giúp giữ thông tin hướng chuyển động.

````python
train_features, train_targets, train_rows = datasets["train"]
example_label = Gesture.SWIPE_LEFT
matching_clips = np.flatnonzero(train_targets == int(example_label))
if len(matching_clips) == 0:
    raise ValueError(f"Tập train chưa có clip {example_label.name}.")
example_index = int(matching_clips[0])
example_row = train_rows[example_index]

with np.load(PROJECT_ROOT / example_row["processed_path"], allow_pickle=False) as archive:
    raw_landmarks = archive["landmarks_xy"]
    frame_timestamps_ms = archive["frame_timestamps_ms"]
    frame_size = (int(archive["frame_width"].item()), int(archive["frame_height"].item()))
    stored_sequence = archive["features"]

print(f"Clip minh họa: {example_row['clip_id']}")
print(f"Nhãn: {example_label.name}")
print(f"Kích thước ảnh: {frame_size}")
print(f"Landmark gốc: {raw_landmarks.shape}")
print(f"Chuỗi đã lưu: {stored_sequence.shape}")
````

````python
display(Code(inspect.getsource(extract_features), language="python"))

first_vector = extract_features(raw_landmarks[0], frame_size)
print(f"Phiên bản đặc trưng: {FEATURE_VERSION}")
print(f"Số đặc trưng/frame: {first_vector.shape}")
print(f"Sáu tọa độ cục bộ đầu: {np.round(first_vector[:6], 4)}")
print(f"Hai tọa độ cổ tay cuối: {np.round(first_vector[-2:], 4)}")
````

#### Đồng bộ thời gian và bộ đệm

`SequenceBuffer` lấy mẫu theo timestamp ở 20 Hz, cách nhau 50 ms. Bộ đệm
giữ 30 bước, tương ứng khoảng 1,5 giây. Mất tay hoặc khoảng hở tracking
quá 100 ms làm khởi tạo lại chuỗi. Khe hở ngắn được nội suy giữa hai mẫu hợp lệ.

Ô dưới phát lại landmark đã lưu qua đúng hai hàm dùng khi thu và chạy webcam,
rồi đối chiếu kết quả với đặc trưng trong tệp NPZ.

````python
buffer = SequenceBuffer(config.sequence_length, config.sample_rate_hz)
rebuilt_sequence = None

for timestamp_ms, landmarks in zip(frame_timestamps_ms, raw_landmarks, strict=True):
    vector = extract_features(landmarks, frame_size)
    candidate = buffer.push(timestamp_ms, vector)
    if candidate is not None:
        rebuilt_sequence = candidate

if rebuilt_sequence is None:
    raise ValueError("Clip minh họa chưa tạo được chuỗi đầy.")
np.testing.assert_allclose(rebuilt_sequence, stored_sequence, rtol=1e-5, atol=1e-6)

print(f"Đầu ra bộ đệm: {rebuilt_sequence.shape}")
print(f"Khoảng thời gian mẫu: {1000 / config.sample_rate_hz:.0f} ms")
print("Phát lại khớp với đặc trưng đã lưu.")
````

````python
sample_time = np.arange(len(rebuilt_sequence)) / config.sample_rate_hz
figure, axes = plt.subplots(1, 2, figsize=(11, 4), layout="constrained")

pixels = raw_landmarks[0] * np.asarray(frame_size)
axes[0].scatter(pixels[:, 0], pixels[:, 1], s=25)
for index, (x, y) in enumerate(pixels):
    axes[0].annotate(str(index), (x, y), xytext=(4, 3), textcoords="offset points", fontsize=9)
axes[0].set_xlim(0, frame_size[0])
axes[0].set_ylim(frame_size[1], 0)
axes[0].set_aspect("equal")
axes[0].set_xlabel("x trong ảnh (pixel)")
axes[0].set_ylabel("y trong ảnh (pixel)")
axes[0].set_title("21 landmark ở frame đầu")

axes[1].plot(sample_time, rebuilt_sequence[:, 42], label="Cổ tay x")
axes[1].plot(sample_time, rebuilt_sequence[:, 43], label="Cổ tay y")
axes[1].set_xlabel("Thời gian từ mẫu đầu (giây)")
axes[1].set_ylabel("Tọa độ / cạnh ảnh dài nhất")
axes[1].set_title(f"Quỹ đạo cổ tay — {example_label.name}")
axes[1].legend()
axes[1].grid(alpha=0.25)
plt.show()
````

### 4.3. Kiến trúc LSTM trong model.py

Luồng tensor là **[batch, 30, 44] → LSTM 1 tầng, hidden size 64 →
đầu ra bước cuối → Linear → 5 logits**.

Cross-entropy dùng trực tiếp logits khi huấn luyện. Khi dự đoán, softmax
chuyển logits thành xác suất, sau đó chọn nhãn có xác suất cao nhất.
Kiến trúc lấy từ lớp `GestureLSTM` dùng chung với ứng dụng.

````python
display(Code(inspect.getsource(GestureLSTM), language="python"))
architecture = GestureLSTM(
    feature_dim=config.feature_dim,
    hidden_size=config.hidden_size,
    num_layers=config.num_layers,
    num_classes=config.num_classes,
)
print(architecture)
print(f"Số tham số: {sum(parameter.numel() for parameter in architecture.parameters()):,}")

with torch.inference_mode():
    sample_input = torch.from_numpy(rebuilt_sequence).unsqueeze(0)
    sample_logits = architecture(sample_input)
print(f"Shape đầu vào: {tuple(sample_input.shape)}")
print(f"Shape logits: {tuple(sample_logits.shape)}")
````

### Dự đoán một clip

Thay `test_clip_index` để xem mẫu khác. `predict_gesture()` thêm chiều batch,

````python
test_features, test_targets, test_rows = datasets["test"]
test_clip_index = 0
if not 0 <= test_clip_index < len(test_targets):
    raise IndexError(f"test_clip_index cần nằm trong khoảng 0 đến {len(test_targets) - 1}.")

sequence = test_features[test_clip_index]
actual = Gesture(int(test_targets[test_clip_index]))
predicted, confidence = predict_gesture(model, sequence)
print(f"Clip: {test_rows[test_clip_index]['clip_id']}")
print(f"Nhãn thật: {actual.name}")
print(f"Dự đoán: {predicted.name}")
print(f"Độ tin cậy: {confidence:.2%}")
````

### 5.3. Đi từng dòng forward

`batch_first=True` đặt batch ở đầu. `sequence[:, -1, :]` giữ mọi mẫu,
chọn bước cuối và lấy đủ 64 giá trị ẩn. Linear biến 64 giá trị thành
năm logits. Ô sau tách đúng các phép tính và đối chiếu với `model(x)`.

````python
with torch.inference_mode():
    batch = torch.from_numpy(sequence).unsqueeze(0)
    recurrent_outputs, recurrent_state = model.lstm(batch)
    last_hidden = recurrent_outputs[:, -1, :]
    traced_logits = model.classifier(last_hidden)
    direct_logits = model(batch)
    torch.testing.assert_close(traced_logits, direct_logits)
    probabilities = traced_logits.softmax(dim=1)[0]

print("NumPy:", sequence.shape)
print("Thêm batch:", tuple(batch.shape))
print("Qua LSTM:", tuple(recurrent_outputs.shape))
print("Lấy bước cuối:", tuple(last_hidden.shape))
print("Qua Linear:", tuple(traced_logits.shape))
print("Forward tách bước khớp với model(batch).")
for name, probability in zip(CLASS_NAMES, probabilities.tolist(), strict=True):
    print(f"{name:12} {probability:.2%}")
````

````python
report = evaluate_model(experiment_config)
print(f"Số clip đúng: {report['correct']}/{report['samples']}")
print(f"Accuracy: {report['accuracy']:.2%}")
print(f"Macro-F1: {report['macro_f1']:.4f}")
print(f"Epoch checkpoint: {report['checkpoint_epoch']}")
print(f"Báo cáo: {experiment_config.output_dir / 'evaluation.json'}")

for name, metrics in report["per_class"].items():
    print(
        f"{name:12} | support={metrics['support']:2} | "
        f"precision={metrics['precision']:.3f} | recall={metrics['recall']:.3f} | "
        f"F1={metrics['f1']:.3f}"
    )
````
